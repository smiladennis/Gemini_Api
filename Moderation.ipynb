```python
import google.generativeai as genai

# Configure your API key
genai.configure(api_key=" ")

# Choose the model
model_name = "gemini-flash-latest"


def get_completion_from_messages(prompt,
                                 model=model_name,
                                 temperature=0,
                                 max_tokens=50):

    model = genai.GenerativeModel(model)

    response = model.generate_content(
        prompt,
        generation_config={
            "temperature": temperature,
            "max_output_tokens": max_tokens
        }
    )

    #reply=response.candidates
    #print("reply : ",reply)
    # If nothing returned
    if not response.candidates:
        return "No response generated."

    candidate = response.candidates[0]

    print("Finish Reason:", candidate.finish_reason)

    # Return text only if it exists
    if hasattr(candidate, "content") and candidate.content.parts:
        return "".join(
            part.text for part in candidate.content.parts
            if hasattr(part, "text")
        )

    return f"No text generated. Finish Reason: {candidate.finish_reason}"


def moderation_check(input_text):

    prompt = f"""
You are a content moderation AI.

Classify the following text.

Reply ONLY with one word.

Safe
or
Flagged

Text:
{input_text}
"""

    return get_completion_from_messages(
        prompt,
        max_tokens=100
    )


# Example
text = """
Here's the plan.
We get the warhead,
and we hold the world ransom...

FOR ONE MILLION DOLLARS!
"""

response = moderation_check(text)
print("Result:", response)
```
